```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("IMDB Dataset.csv")
```

```python
df["review"] = df["review"].str.lower()
```

## Remove HTML Tags

```python
import re

def remove_html_tags(text):
    pattern = re.compile("<.+?>")
    return pattern.sub(r"", text)
```

```python
df["review"] = df["review"].apply(remove_html_tags)
```

## Remove URLS

```python
def remove_urls(text):
    pattern = re.compile(r'(?:https?:\/\/|ftp:\/\/|www\.)[-a-zA-Z0-9@:%._\+~#=]{1,256}\.[a-zA-Z0-9()]{1,63}\b(?:[-a-zA-Z0-9()@:%_\+.~#?&\/=]*)')
    return pattern.sub(r"", text).strip()
```

```python
df["review"] = df["review"].apply(remove_urls)
```

## Remove Punctuation

```python
import string
from string import punctuation

def remove_punctuation(text):
    translator = str.maketrans("", "", punctuation)
    return text.translate(translator)
    
```

```python
df["review"] = df["review"].apply(remove_punctuation)
```

## Remove Stop words

```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords', quiet=True)
STOP_WORDS = set(stopwords.words("english"))

_PATTERN = r'\b(?:' + '|'.join(re.escape(word) for word in sorted(STOP_WORDS, key=len, reverse=True)) + r')\b'
STOPWORD_REGEX = re.compile(_PATTERN, flags=re.IGNORECASE)
SPACE_REGEX = re.compile(r'\s+')

def remove_stopwords_from_string(text):
    """Removes stopwords from text using pre-compiled regex."""
    if not text:
        return ""
        
    # 3. Use the pre-compiled regex engine for substitution
    cleaned_text = STOPWORD_REGEX.sub('', text)
    
    # 4. Use pre-compiled whitespace regex
    cleaned_text = SPACE_REGEX.sub(' ', cleaned_text).strip()
    
    return cleaned_text

```

```python
df["review"] = df["review"].apply(remove_stopwords_from_string)
```

## Tokenization

```python
import nltk 
from nltk.tokenize import word_tokenize, sent_tokenize

# Download both required data packages for modern NLTK versions
nltk.download("punkt")
nltk.download("punkt_tab")

def tokenize(text):
    word_tokens = word_tokenize(text)
    
    return word_tokens
```

```python
df["review"] = df["review"].apply(tokenize)
```
