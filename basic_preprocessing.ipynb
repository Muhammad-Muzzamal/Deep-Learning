```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("IMDB Dataset.csv")
```

```python
df["review"] = df["review"].str.lower()
```

## Remove HTML Tags

```python
import re

def remove_html_tags(text):
    pattern = re.compile("<.+?>")
    return pattern.sub(r"", text)
```

```python
df["review"] = df["review"].apply(remove_html_tags)
```

## Remove URLS

```python
def remove_urls(text):
    pattern = re.compile(r'(?:https?:\/\/|ftp:\/\/|www\.)[-a-zA-Z0-9@:%._\+~#=]{1,256}\.[a-zA-Z0-9()]{1,63}\b(?:[-a-zA-Z0-9()@:%_\+.~#?&\/=]*)')
    return pattern.sub(r"", text).strip()
```

```python
df["review"] = df["review"].apply(remove_urls)
```

## Remove Punctuation

```python
import string
from string import punctuation

def remove_punctuation(text):
    translator = str.maketrans("", "", punctuation)
    return text.translate(translator)
    
```

```python
df["review"].apply(remove_punctuation)
```

## Remove Stop words

```python
import re
from nltk.corpus import stopwords

nltk.download('stopwords')

def remove_stopwords_from_string(text):
    stop_words = set(stopwords.words('english'))
    
    # Build a regex pattern matching any stopword as a whole word, case-insensitively
    pattern = r'\b(?:' + '|'.join(re.escape(word) for word in stop_words) + r')\b'
    
    # Replace stopwords with an empty string
    cleaned_text = re.sub(pattern, '', text, flags=re.IGNORECASE)
    
    # Clean up any leftover double spaces
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
    
    return cleaned_text

```

```python
df["review"] = df["review"].apply(remove_stopwords_from_string)
```

## Tokenization

```python
import nltk 
from nltk.tokenize import word_tokenize, sent_tokenize

# Download both required data packages for modern NLTK versions
nltk.download("punkt")
nltk.download("punkt_tab")

def tokenize(text):
    word_tokens = word_tokenize(text)
    
    return word_tokens
```

```python
df["review"] = df["review"].apply(tokenize)
```
